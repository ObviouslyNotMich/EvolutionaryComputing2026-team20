## Imports

```python
from pathlib import Path
from typing import Literal
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

from data_analysis.database_ops import merge_individual_databases
from data_analysis.plots import plot_fitness

PROJECT_ROOT = Path.cwd().parent.parent
DATA_PATH = PROJECT_ROOT.joinpath("__data__")

ONE_PATH = DATA_PATH.joinpath("one_results")
N_PATH = DATA_PATH.joinpath("n_results")
RANDOM_PATH = DATA_PATH.joinpath("random_results")

# Define sources and combined locations
one_sources = sorted(ONE_PATH.glob("*.db"))
n_sources = sorted(N_PATH.glob("*.db"))
random_sources = sorted(RANDOM_PATH.glob("*.db"))

one_combined = DATA_PATH / "one_combined.db"
n_combined = DATA_PATH / "n_combined.db"
random_combined = DATA_PATH / "random_combined.db"
```

## Merging Databases

This only needs to be run once to merge all runs into a single file.

```python
merge_individual_databases(
    source_paths=one_sources,
    target_path=one_combined
)

merge_individual_databases(
    source_paths=n_sources,
    target_path=n_combined
)

merge_individual_databases(
    source_paths=random_sources,
    target_path=random_combined
)
```

## Plots for data analysis
